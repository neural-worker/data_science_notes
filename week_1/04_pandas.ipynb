```python
df_titanic = fetch_openml('titanic', version=1, as_frame=True, parser='auto').frame
```

```python
df_titanic['Title'] = df_titanic['name'].str.extract(r',\s([^\.]+)\.')
```

```python
title_status = pd.DataFrame({
    'Title': ['Mr', 'Mrs', 'Miss', 'Master'],
    'status': ['Adult male', 'Married female', 'Single female', 'Child male']
})
```

```python
df_titanic = pd.merge(df_titanic, title_status, how='left', on='Title')
```

```python
df_titanic['status'] = df_titanic['status'].fillna('Rare')
```
