Is it true that passengers who travelled only parents and children survived more frequently than those who travelled only siblings and spouses?

```python
df['had_only_sibsp'] = (df['sibsp'] > 0) & (df['parch'] == 0)
df['had_only_parch'] = (df['sibsp'] == 0) & (df['parch'] > 0)
df['had_both'] = (df['sibsp'] > 0) & (df['parch'] > 0)
df.groupby(['had_only_sibsp', 'had_only_parch', 'had_both'], as_index=False)['survived'].mean()
```

```python
def get_family_structure(row):
    if row['sibsp'] == 0 and row['parch'] == 0:
        return 'Alone'
    elif row['sibsp'] > 0 and row['parch'] == 0:
        return 'Only SibSp'
    elif row['sibsp'] == 0 and row['parch'] > 0:
        return 'Only Parch'
    else:
        return 'Both'

df['family_structure'] = df.apply(get_family_structure, axis=1)
df.groupby('family_structure')['survived'].agg(['count', 'mean'])
```
