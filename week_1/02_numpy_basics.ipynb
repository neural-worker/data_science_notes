```python
a = np.ones((5, 5))
a[:, [0,-1]] = 0
a[[0, -1], :] = 0
print(a.size)
# or !!!!!!!!!!!!!!!!!!!!!!!!!!
a = np.ones((5, 5))
a = np.pad(a, pad_width=1, mode='constant', constant_values=0)
print(a.size)
```

```python
a = np.arange(1, 61, 2).reshape((5,6))
index = 10
print(a)
print(np.unravel_index(index-1, a.shape))
```

```python
a = np.arange(0, 336).reshape((6, 7, 8))
elem = 100
print(np.unravel_index(elem-1, a.shape))
```

```python
np.tile(np.array([[1, 0], [0, 1]]), (4, 4))
```

```python
a = np.arange(9, -36, -5, dtype='float64').reshape((3, 3))
norma = np.sum(a, axis=1)[:, np.newaxis]
a /= norma
np.sum(a, axis=1)[:, None]
```

```python
color = np.dtype([('red', np.uint8),
                ('green', np.uint8),
                ('blue', np.uint8),
                ('alpha', np.uint8)])
point = np.array([(0, 0, 0, 0), (255, 255, 255, 50)], dtype=color)
point['red']
```

```python
a = np.arange(15).reshape((5,3))
b = np.arange(6).reshape((3, 2))
a.dot(b)
```

```python
a = np.arange(-15, 15).reshape((5, 6))
mask = (3 <= a) & (a <= 8)
a[mask] *= -1
a
```
