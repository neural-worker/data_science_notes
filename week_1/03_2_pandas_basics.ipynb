```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df = sns.load_dataset('titanic')
```

```python
df.index = np.arange(1, len(df)+1)
```

```python
df
```

```python
df[df['who']=='child']['age'].mean()
```

```python
df.groupby(['who', 'pclass', 'age'])['survived'].agg(['mean'])
```

```python
df.groupby('class', as_index=False).agg(
    passengers=('survived', 'count'),
    avg_age=('age', 'mean'),
    max_fare=('fare', 'max')
)
```

```python
df.groupby(['who', 'class'])['survived'].mean()
```

```python
df.groupby('survived')['age'].mean()
```

```python
df['class'].value_counts(normalize=True) * 100
```

```python
pd.crosstab(df['sex'], df['survived'], normalize='index') * 100
```
